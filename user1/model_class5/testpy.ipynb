```python
import fiftyone as fo
import fiftyone.zoo as foz

# 1. FiftyOne Zoo에서 FER2013 데이터셋 로드
# FER2013은 기본적으로 얼굴만 크롭된 이미지와 감정 라벨을 포함합니다.
dataset = foz.load_zoo_dataset("fer2013", split="test")

# 2. 데이터셋에 포함된 라벨 종류(Class) 확인
# 결과 예: ['anger', 'disgust', 'fear', 'happy', 'sad', 'surprise', 'neutral']
print(f"✅ 분류된 라벨 종류: {dataset.default_classes}")

# 3. 각 라벨별 데이터 개수 통계 확인
stats = dataset.count_values("ground_truth.label")
print("\n📊 라벨별 이미지 개수:")
for label, count in stats.items():
    print(f"- {label}: {count}장")

# 4. 시각화 툴(App) 실행하여 눈으로 직접 확인
session = fo.launch_app(dataset)
```

```python
import fiftyone as fo
import fiftyone.zoo as foz

# 1. 데이터 로드 (이전과 동일)
dataset = foz.load_zoo_dataset(
    "open-images-v7",
    split="validation",
    label_types=["detections"],
    classes=["Human face"],
    max_samples=2000,
    drop_existing_dataset=True
)

# 2. 필터링 부분 수정 (fo.ViewField 사용)
F = fo.ViewField
face_view = dataset.filter_labels("ground_truth", F("label") == "Human face")

# 3. 속성 통계 확인 (동일)
print("📊 얼굴 데이터 속성 통계:")
for attr in ["IsOccluded", "IsTruncated", "IsGroupOf"]:
    # 속성 값이 존재하는지 확인하는 통계
    count = face_view.count_values(f"ground_truth.detections.attributes.{attr}.value")
    print(f"- {attr}: {count}")

# 4. 앱 실행
session = fo.launch_app(dataset)
```
