```python
import torch
import os

# 확인하고 싶은 5개 파일의 정확한 경로를 아래 리스트에 넣으세요.
pt_files = [
    '/workspace/user3/yolo26m.pt',
    '/workspace/user3/yolo26n.pt',
    '/workspace/user3/yolo26s-seg.pt',
    '/workspace/user3/yolo26s.pt',
    '/workspace/user3/yolov8n-seg.pt'
    # ... 나머지 파일 경로 추가
]

print("=== PT 파일 정밀 분석 시작 ===\n")

for file_path in pt_files:
    if not os.path.exists(file_path):
        print(f"❌ 파일을 찾을 수 없습니다: {file_path}")
        continue
        
    print(f"📂 분석 중: {os.path.basename(file_path)}")
    print(f"   - 경로: {file_path}")
    print(f"   - 용량: {os.path.getsize(file_path) / (1024*1024):.2f} MB")
    
    try:
        # map_location='cpu'로 안전하게 로드
        data = torch.load(file_path, map_location='cpu')
        
        if isinstance(data, dict):
            keys = list(data.keys())
            print(f"   - 최상위 구조(Keys): {keys}")
            
            # YOLOv8 모델의 전형적인 특징 검사
            if 'model' in data:
                model_obj = data['model']
                print(f"   - 모델 객체 타입: {type(model_obj).__name__}")
                
                # 클래스 이름(names)이 저장되어 있는지 확인
                if hasattr(model_obj, 'names'):
                    print(f"   - ⭐ 학습된 클래스명: {model_obj.names}")
            
            # 마지막 학습 Epoch 확인
            if 'epoch' in data:
                print(f"   - 마지막 학습 Epoch: {data['epoch']}")
                
        else:
            print("   - 단순 텐서(Tensor) 혹은 알 수 없는 구조입니다.")
            
    except Exception as e:
        print(f"   - ❌ 로드 실패 (파일 손상 또는 호환성 문제): {e}")
    
    print("-" * 50)
```
