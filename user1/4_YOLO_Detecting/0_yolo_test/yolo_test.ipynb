```python
import os
from ultralytics import YOLO

def detect_faces(image_path, model_path='/workspace/user3/yolo26n.pt'):
    # 1. 모델과 이미지가 존재하는지 사전 검사
    if not os.path.exists(model_path):
        print(f"❌ 오류: 모델 가중치 파일을 찾을 수 없습니다. 경로를 확인하세요: {model_path}")
        return
    if not os.path.exists(image_path):
        print(f"❌ 오류: 테스트 이미지를 찾을 수 없습니다. 경로를 확인하세요: {image_path}")
        return

    print(f"🔍 '{os.path.basename(image_path)}' 이미지 분석을 시작합니다...")

    # 2. YOLOv8 '얼굴 전용' 가중치 로드
    model = YOLO(model_path)

    # 3. 인퍼런스 (추론) 실행
    # conf=0.3: 확신도가 30% 이상인 객체(얼굴)만 박스 처리 (오탐지 방지)
    # iou=0.5: 겹치는 박스들 중 가장 정확한 것 하나만 남김 (중복 탐지 방지)
    # save=True: 박스가 그려진 최종 결과 이미지를 자동 저장
    results = model.predict(source=image_path, save=True, conf=0.3, iou=0.5)

    # 4. 결과 출력
    for r in results:
        face_count = len(r.boxes)
        print(f"✅ 탐지 완료! 총 {face_count}명의 얼굴을 찾았습니다.")
        print(f"📁 결과 이미지 저장 경로: {r.save_dir}")

if __name__ == "__main__":
    # 테스트할 이미지 경로를 여기에 적어주세요.
    test_image = '/workspace/user1/4_YOLO_Detectng/0_yolo_test/test_pic/2026-04-20_4.01.10---.png' 
    detect_faces(test_image)
```
